```python
import plistlib

import pandas as pd

with open("rawdata_am/Library_kha.xml", "rb") as f:
    data = plistlib.load(f)

tracks = data["Tracks"]

df = pd.DataFrame.from_dict(tracks, orient="index")


print(df.shape)
print(df.columns.tolist())
```

```python
integer_columns = [
    "bit_rate",
    "sample_rate",
    "play_count",
    "skip_count",
    "normalization",
    "file_folder_count",
    "library_folder_count",
    "disc_number",
    "disc_count",
    "track_number",
    "track_count",
    "year",
    "artwork_count",
    "movement_number",
    "movement_count",
    "rating",
    "album_rating"
]

for column in integer_columns:
    df[column] = df[column].astype("Int64")
```
