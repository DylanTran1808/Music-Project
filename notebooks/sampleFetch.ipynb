```python
import requests

params = {
    "term": "Xương rồng",
    "media": "music",
    "entity": "song",
    "country": "us",
    "limit": 20,
}

response = requests.get(
    "https://itunes.apple.com/search",
    params=params,
    timeout=10,
)

response.raise_for_status()

data = response.json()

print("Number of results:", data["resultCount"])

for result in data["results"]:
    print(
        result.get("trackName"),
        "-",
        result.get("artistName"),
        "|",
        result.get("previewUrl"),
    )
```
